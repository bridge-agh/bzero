```python
import matplotlib.pyplot as plt
import numpy as np

# Provided data
data = [
    0.48619598, 0.42365015, 0.3828488, 0.35133156, 0.31704935, 0.29189056,
    0.26599902, 0.22007817, 0.18466048, 0.15189256, 0.12817383, 0.11010742,
    0.095458984, 0.075927734, 0.05102539, 0.047607422, 0.036621094, 0.028808594,
    0.011474609, 0.0009765625
]

# Generating x values from 0.0 to 1.0 with a step of 0.05
x_values = np.linspace(0.0, 1.0, 20)

print(len(data))

# Plotting
plt.figure(figsize=(8, 5))

# plot y line

# these two little shifted to be visible close
plt.axhline(0.2790, color='tab:red', linestyle='--', label='bzero winrate againts DDS')
plt.axhline(0.276, color='tab:green', linestyle='--', label='brl-from-scratch-rl winrate againts DDS')

plt.axhline(0.31545, color='tab:purple', linestyle='--', label='bzero-xl winrate againts DDS')


# data
plt.plot(x_values, data, marker='o', linestyle='-', color='tab:blue')

# Adding labels and title
plt.xlabel('misschance', fontsize=12)
plt.ylabel('winrate', fontsize=12)
plt.grid()
plt.legend()

# x ticks every 0.05
plt.xticks(np.arange(0, 1.05, step=0.1))

# Displaying the plot
plt.savefig("research_out/misschance_win.pdf", dpi=300, bbox_inches='tight')
```
